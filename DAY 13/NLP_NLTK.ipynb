### **Topic 6: The NLP Pipeline**

This is an important concept that shows how raw text is converted into a format a machine can understand.

**The standard flow:**

1.  **Raw Text**: The input data.
2.  **Tokenization**: Breaking text into words/sentences.
3.  **Cleaning**: Removing noise (special characters, etc.).
4.  **Stopword Removal**: Removing common words like 'is', 'the', 'and'.
5.  **Stemming / Lemmatization**: Reducing words to their root form.
6.  **Feature Extraction**: Converting text into numbers.
7.  **Machine Learning Model**: Training the algorithm.
8.  **Prediction**: Getting the final output.

```mermaid
graph TD
    A[Raw Text] --> B[Tokenization]
    B --> C[Cleaning]
    C --> D[Stopword Removal]
    D --> E[Stemming / Lemmatization]
    E --> F[Feature Extraction]
    F --> G[Machine Learning Model]
    G --> H[Prediction]
```

## **Module 2 — Text Preprocessing**

Now we start the most important beginner part of NLP: cleaning text before giving it to a machine-learning model.

### **Topic 1 — Lowercasing**

Computers are case-sensitive, meaning they treat these as different words:
*   Python
*   python
*   PYTHON

Lowercasing ensures the model treats them all the same.

##### --> **LOWER THE TEXT**

````python
text = "Learning NLP is FUN and Python is Powerful!"

# Convert the entire string to lowercase
lower_text = text.lower()

print("Original:", text)
print("Lowercased:", lower_text)
````

### **Topic 2 — Removing Punctuation**

We usually don't need symbols like: `. , ! ? :` because they often don't help the machine understand the meaning of the words.

Python provides a string of all punctuation marks in `string.punctuation`.

````python
import string

text = "Hello! How are you? I'm learning NLP."

# string.punctuation contains: !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
# We can use a loop or translation table to remove them

#'string.punctuation' a python library which contain the puntuation
clean_text = ""
for char in text:
    if char not in string.punctuation:
        clean_text += char

#OR

"""text = text.translate(
    str.maketrans("", "", string.punctuation)
)"""

print("Original:", text)
print("Cleaned:", clean_text)
````

### **Topic 3 — Stopwords**

Stopwords are very common words that often don't provide much useful information for understanding the unique meaning of a sentence.

Examples include: `the`, `is`, `am`, `a`, `an`, `and`, `in`, `of`.

NLTK has a built-in list of stopwords for many languages.

````python
import nltk
# Uncomment the line below if you get a LookupError!
# nltk.download('stopwords')
from nltk.corpus import stopwords

text1 = "I am learning NLP and it is very interesting"

# Get the words in English stopwords
stop_words = set(stopwords.words('english'))

# Split the sentence into individual words
x = nltk.word_tokenize(text1)

# Filter out the stopwords from 'x'
filtered_words = [
    word for word in x
    if word.lower() not in stop_words
]

print(filtered_words)
````

### **Topic 4 - Complete Basic Preprocessing**

````python
import nltk
import string

#nltk.download("stopwords")

from nltk.corpus import stopwords

text = "I am learning NLP, and NLP is very interesting!"

# Step 1: Convert text to lowercase
text = text.lower()

# Step 2: Remove punctuation
text = text.translate(
    str.maketrans("", "", string.punctuation)
)

# Step 3: Split text into words
words = text.split()

# Step 4: Get English stopwords
stop_words = set(stopwords.words("english"))

# Step 5: Remove stopwords
clean_words = [
    word for word in words
    if word not in stop_words
]

print(clean_words)
````

## **Module 3 — Stemming & Lemmatization**

Now we learn how NLP handles different forms of the same word.

### **Topic 1 — Stemming**

Stemming is the process of reducing a word to its word stem (root form). For example, "playing", "played", and "plays" all have the root **"play"**.

NLTK uses the **PorterStemmer** algorithm for this.

````python
from nltk.stem  import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "played", "plays", "studying", "studies"]
stem_words = [
    stemmer.stem(word) for word in words
]

print(f'Actual Words: {words}')
print()
print(f"Stem_words : {stem_words}")
````

<br>
<br>

### **Topic 2 — Lemmatization**

**Lemmatization** tries to return the actual dictionary form of a word.

Unlike stemming, it uses a vocabulary and morphological analysis of words to return the base form, known as the **lemma**.

**Examples:**
*   running → **run**
*   studies → **study**
*   better → **good**

````python
import nltk

# Download the WordNet dictionary
#nltk.download("wordnet")

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["playing", "played", "studies", "cars"]

for word in words:
    # lemmatize() tries to find the dictionary/base form
    result = lemmatizer.lemmatize(word)

    print(word, "→", result)
````

<br>
<br>

<br>
<br>

### **Topic 3 — Stemming vs Lemmatization**

| Feature | Stemming | Lemmatization |
| :--- | :--- | :--- |
| **Speed** | Faster | Slower |
| **Intelligence** | Simple | More intelligent |
| **Correctness** | May produce incorrect words | Usually produces valid words |
| **Method** | Removes word endings | Uses dictionary knowledge |

#### **Example Comparison:**

**Stemming:**  
`studying` → **studi**

**Lemmatization:**  
`studying` → **studying**

**Why did that happen?**

Because lemmatization needs to know the **Part of Speech (POS)** to work properly. By default, it assumes every word is a **Noun**. Since "studying" is a verb, the lemmatizer didn't change it. We'll learn how to handle POS tags soon!

<br>
<br>

### **Topic 4 - Lemmanization in Pos**

````python
import nltk

#nltk.download("wordnet")

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

word = "running"

# pos="v" tells NLTK that the word is a verb
result = lemmatizer.lemmatize(word, pos="v")

print(result)
````

### **Module 4 — POS Tagging**
#### **Topic 1 — What is POS Tagging?**

POS = Part of Speech

It tells us what role each word plays in a sentence.

Example:

I love Python.

I       → Pronoun
love    → Verb
Python  → Noun

<br>
<br>

#### **Topic 2 — POS Tagging with NLTK**

````python
import nltk

# Download the data required for POS tagging
#nltk.download("averaged_perceptron_tagger_eng")
#nltk.download("punkt_tab")

from nltk.tokenize import word_tokenize
from nltk import pos_tag

text = "I love Python. Python is easy to learn."

# Convert the sentence into individual words
words = word_tokenize(text)

# Identify the grammatical role of each word
tags = pos_tag(words)

print(tags)
````

<br>
<br>

### **Topic 3 — Understanding the Tags**

You don't need to memorize all tags now. Just understand the common ones:

| Tag | Description | Example |
| :--- | :--- | :--- |
| **NN** | Noun, singular | car |
| **NNP** | Proper Noun, singular | **Python** |
| **VB** | Verb, base form | learn |
| **VBP** | Verb, present | **love** |
| **VBG** | Verb, gerund (-ing) | running |
| **JJ** | Adjective | **great** |
| **RB** | Adverb | **quickly** |
| **PRP** | Personal Pronoun | **I**, he, she |

**Quick Example:**
*   Python → **NNP**
*   love → **VBP**
*   great → **JJ**
*   quickly → **RB**

<br>
<br>

#### **Topic 4 - See Pos Clearly**

````python
import nltk

#nltk.download("averaged_perceptron_tagger_eng")

from nltk.tokenize import word_tokenize
from nltk import pos_tag

text = "Python is an amazing programming language."

# Break the sentence into words
words = word_tokenize(text)

# Assign a POS tag to every word
tags = pos_tag(words)

# Print each word and its POS tag
for word, tag in tags:
    print(word, "→", tag)
````

<br>
<br>

### **Topic 5 — Why POS Tagging Matters**

POS tagging becomes useful when we want to understand how a word is being used in context.

**For example:**

1.  **"I am running."**  
    *   `running` → **Verb**

2.  **"I bought running shoes."**  
    *   `running` → **Adjective**

Same word, different role. POS tagging helps NLP systems understand this difference, which is crucial for tasks like translation and sentiment analysis.

## **Module 5 — Named Entity Recognition (NER)**

### **Topic 1 — What is NER?**

NER is the process of finding and classifying specific entities in text.

**Common Entity Types:**
*   **PERSON**: People, including fictional characters.
*   **ORGANIZATION**: Companies, agencies, institutions.
*   **GPE (Geo-Political Entity)**: Countries, cities, states.
*   **DATE**: Days, months, years.

<br>
<br>

````python
import nltk

nltk.download("maxent_ne_chunker_tab")
nltk.download("words")

from nltk.tokenize import word_tokenize
from nltk import pos_tag, ne_chunk

test = "Elon Musk founded SpaceX in the United States."

#STEP 1
w = word_tokenize(test)

#STEP 2
tags = pos_tag(w)

#STEP 3
entities = ne_chunk(tags)

print(entities)
````

<br>
<br>

### **Topic 3 — Understanding the Process**

Notice the pipeline:

```
Raw Text
   ↓
Tokenization
   ↓
POS Tagging
   ↓
NER (ne_chunk)
   ↓
Entities Identified
```

**Why POS Tagging first?**  
NER systems use Part-of-Speech tags to find patterns. For example, a sequence of Proper Nouns (`NNP`) often signals a name or a place.

<br>
<br>

### **Topic 4 - Real Example**

````python
import nltk

#nltk.download("maxent_ne_chunker_tab")
#nltk.download("words")

from nltk.tokenize import word_tokenize
from nltk import pos_tag, ne_chunk

text = "Microsoft was founded by Bill Gates in the United States."

# Convert text into individual words
words = word_tokenize(text)

# Assign POS tags to the words
tags = pos_tag(words)

# Find named entities such as people and organizations
entities = ne_chunk(tags)

print(entities)
````

### **Topic 5 — Why NER is Important**

NER is used in various industries to turn unstructured text into structured data:

*   **Chatbots**: To understand who or what a user is talking about.
*   **Search Engines**: To provide relevant results for specific entities (e.g., searching for "Tesla" as a company vs. "Tesla" as a scientist).
*   **Financial Analysis**: Extracting numbers and company names from news reports.
*   **Resume Parsing**: Automatically identifying skills, universities, and past employers.

## **Module 6 — Text Representation**

Computers and Machine Learning models cannot directly understand human language like "I love Python". They can only perform calculations on **numbers**.

To bridge this gap, we convert:
**Text → Numbers (Vectors)**

<br>
<br>

### **Topic 1 — Bag of Words (BoW)**

Bag of Words (BoW) counts how many times each word appears in a document.

**Example:**
1. "I love Python"
2. "I love NLP"

**Vocabulary:** `[I, love, Python, NLP]`

**Representation:**
*   "I love Python" → `[1, 1, 1, 0]`
*   "I love NLP"    → `[1, 1, 0, 1]`

<br>
<br>

### **Topic 3 - CounterVectorizer**

````python
from sklearn.feature_extraction.text import CountVectorizer

sentence = [
    "I Love Python",
    "I Love Nlp"
]

vectorizer = CountVectorizer()

x = vectorizer.fit_transform(sentence)

# Show the words learned by the vectorizer
#USE --> 'get_features_name_out()'
print(vectorizer.get_feature_names_out())

# Show the numerical representation
print(x.toarray())
````

<br>
<br>

### **Topic 3 — Understanding the Output**

When we run the `CountVectorizer`, it creates a **Vocabulary** based on all the unique words it finds (usually in alphabetical order).

**Our Vocabulary:**
`love` | `nlp` | `python`

**First sentence: "I Love Python"**
*   Contains 'love' (1)
*   Contains 'nlp' (0)
*   Contains 'python' (1)
*   **Vector:** `[1, 0, 1]`

**Second sentence: "I Love Nlp"**
*   Contains 'love' (1)
*   Contains 'nlp' (1)
*   Contains 'python' (0)
*   **Vector:** `[1, 1, 0]`

Notice that the vectorizer automatically lowercased the words and converted them into a grid of numbers. Your model can now do math on these vectors!

<br>
<br>

### **Topic 4 — TF-IDF**

**TF-IDF** stands for **Term Frequency - Inverse Document Frequency**.

Unlike Bag of Words, which only counts how many times a word appears, TF-IDF measures how **important** a word is to a document in a collection.

*   **Term Frequency (TF)**: How often a word appears in a single document.
*   **Inverse Document Frequency (IDF)**: How rare a word is across all documents. Rare words get higher scores!

**Example:**
1. "The movie was good"
2. "The movie was excellent"
3. "The movie was amazing"

Words like "the", "movie", and "was" appear in every sentence, so they get a **low score**. Words like "good", "excellent", and "amazing" are unique to their sentences, so they get a **high score**.

<br>
<br>

### **Topic 5 — Understanding TF-IDF Scores**

In the output above, look at the scores for each word:

*   **'the', 'movie', 'was'**: These appear in all 3 sentences. Because they are common, their TF-IDF score is lower (~0.41).
*   **'amazing', 'excellent', 'good'**: These are unique to their specific sentences. Because they are rare in the collection, they get a much higher score (~0.69).

This helps a machine-learning model focus on the words that actually distinguish one sentence from another!

````python
from sklearn.feature_extraction.text import TfidfVectorizer

sentences = [
    "I love Python",
    "I love NLP",
    "Python is powerful"
]

# Create the TF-IDF converter
vectorizer = TfidfVectorizer()

# Learn the vocabulary and convert sentences into numerical features
X = vectorizer.fit_transform(sentences)

# Show the words learned
print(vectorizer.get_feature_names_out())

# Show the TF-IDF numerical representation
print(X.toarray())
````

## **Module 7 — NLP + Machine Learning**

Now we connect NLP with Machine Learning.

The basic workflow is:

**Text** → **Clean** → **Convert to Numbers** → **Train ML Model** → **Prediction**

<br>
<br>

### **Topic 1 - Text Classification**

Text classification means putting text into a category.

**Example:**

> "I won a free lottery!" → **SPAM**
>
> "Can we meet tomorrow?" → **NOT SPAM**

<br>
<br>

### **Topic 2 - Creating Data**

````python
texts = [
    "Win a free lottery",
    "Congratulations you won a prize",
    "Claim your free money",
    "Can we meet tomorrow",
    "I will call you later",
    "Let's study Python"
]

# 1 = Spam
# 0 = Not Spam
labels = [1, 1, 1, 0, 0, 0]
````

<br>
<br>

### **Topic 3 - Convert text into Numbers**

````python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

print(vectorizer.get_feature_names_out())
print(X.toarray())
````

<br>
<br>

### **Topic 4 - Train a Machine Learning Model**

````python
from sklearn.linear_model import LinearRegression

# Create and train a linear regression model
model = LinearRegression()
model.fit(X, labels)
````

<br>
<br>

### **Topic 5 - Make predctions**

````python
new_text = ["You won a free prize"]

# Convert the new sentence using the SAME vectorizer
new_text_vector = vectorizer.transform(new_text)

# Ask the trained model to predict the category
prediction = model.predict(new_text_vector)

print(prediction)
````

<br>
<br>

### **Topic 6 - Complete Mini Example**

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

texts = [
    "Win a free lottery",
    "Congratulations you won a prize",
    "Claim your free money",
    "Can we meet tomorrow",
    "I will call you later",
    "Let's study Python"
]

labels = [1, 1, 1, 0, 0, 0]

# Convert text into numerical TF-IDF features
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)

# Create and train the ML model
model = LogisticRegression()
model.fit(X, labels)

# New text that the model has never seen
new_text = ["Congratulations you won free money"]

# Convert the new text into the same numerical format
new_X = vectorizer.transform(new_text)

# Predict whether it is spam or not
prediction = model.predict(new_X)

if prediction[0] == 1:
    print("SPAM")
else:
    print("NOT SPAM")
````
