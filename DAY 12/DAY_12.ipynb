## Part 1: What is Deep Learning?

**Deep Learning** is a specialized subset of Machine Learning that uses multi-layered **Artificial Neural Networks** to model and solve complex patterns in data.

### Concept Origins
Deep Learning is inspired by the biological structure of the **human brain**, mimicking the way interconnected neurons process information.

### Machine Learning vs. Deep Learning

| Feature | Machine Learning | Deep Learning |
| :--- | :--- | :--- |
| **Model Architecture** | Shallow (few layers) | Deep (many hidden layers) |
| **Feature Extraction** | Manual (requires experts) | Automated (learned from data) |
| **Data Format** | Structured (CSV, SQL) | Unstructured (Images, Audio, Text) |
| **Scaling** | Performance plateaus | Performance improves with more data |

### Real-World Applications
*   **Generative AI:** Large Language Models like GPT.
*   **Computer Vision:** Autonomous vehicles and facial recognition.
*   **Speech Processing:** Voice assistants and real-time translation.
*   **Recommender Systems:** Personalized content on streaming platforms.

> **Key Takeaway:** Deep Learning excels by automatically learning features from vast amounts of unstructured data using complex, multi-layered architectures.

## Part 2: Artificial Neural Networks (ANN)

An **Artificial Neural Network (ANN)** is the structural foundation of deep learning. It consists of thousands or millions of simple processing nodes called **neurons** organized into layers.

### The Architecture of an ANN
1.  **Input Layer:** The entry point for raw data features.
2.  **Hidden Layers:** Where the 'learning' happens through mathematical transformations.
3.  **Output Layer:** The final layer that produces the prediction (e.g., a probability or a numerical value).

### Example: Credit Risk Assessment
*   **Inputs:** Credit score, debt-to-income ratio, employment history.
*   **Hidden Processing:** The network identifies non-linear correlations between these factors.
*   **Output:** A percentage chance of loan default.

> **Key Takeaway:** ANNs transform raw input data into meaningful outputs by passing signals through layers of interconnected computational units.

## Part 3: What is a Neuron?

The **neuron** is the fundamental building block of a neural network. Each neuron receives a signal, processes it, and decides whether to pass it along to the next layer.

### The Mathematical Equation
The output of a single neuron is defined by the following transformation:

$$y = f(\sum_{i=1}^{n} w_i x_i + b)$$

**Components:**
*   $x$: The input data.
*   $w$ (**Weight**): The strength of the connection for that input.
*   $b$ (**Bias**): A constant that allows the model to shift the activation function.
*   $f$ (**Activation Function**): Determines the final output based on the weighted sum.

> **Key Takeaway:** A neuron acts as a mathematical gatekeeper, weighting inputs and applying a transformation to generate a single output signal.

<a name="part-4"></a>
## Part 4: Weights and Bias

Weights and biases are the **learnable parameters** of a neural network. These are the values the model adjusts during training.

### Weights ($w$)
Weights represent the **significance** of a feature. If a specific input is highly predictive of the output, its associated weight will increase during training.

### Bias ($b$)
Bias provides the model with **flexibility**. Mathematically, it behaves like the intercept in a linear equation, allowing the activation function to shift and better fit the underlying data trends.

> **Key Takeaway:** Training a model is the process of finding the optimal combination of weights and biases that produce the most accurate predictions.

<a name="part-5"></a>
## Part 5: Activation Functions

Activation functions introduce **non-linearity**. Without them, a neural network—no matter how deep—would just behave like a simple linear regression model.

### Standard Activation Functions

| Function | Formula | Output Range | Best Use Case |
| :--- | :--- | :--- | :--- |
| **ReLU** | $f(x) = \max(0, x)$ | $[0, \infty)$ | Hidden layers (Standard) |
| **Sigmoid** | $\sigma(x) = \frac{1}{1 + e^{-x}}$ | $(0, 1)$ | Binary Classification |
| **Tanh** | $\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$ | $(-1, 1)$ | Zero-centered hidden layers |

> **Key Takeaway:** Activation functions allow networks to learn complex, non-linear relationships within the data, which is essential for solving real-world problems.

<a name="part-6"></a>
## Part 6: Forward Propagation

**Forward Propagation** is the process where input data is fed through the network to generate a prediction.

### The Sequence of Events
1.  **Input:** Features are fed into the input layer.
2.  **Dot Product:** Each layer calculates the weighted sum of its inputs.
3.  **Activation:** The result is passed through an activation function.
4.  **Propagation:** This output becomes the input for the next layer until the output layer is reached.

> **Key Takeaway:** Forward propagation is the 'thinking' phase where the model applies its current knowledge to make an inference.

<a name="part-7"></a>
## Part 7: Loss Functions

A **Loss Function** measures the error between the model's prediction $(\hat{y})$ and the actual ground truth $(y)$.

### Regression: Mean Squared Error (MSE)
MSE is commonly used for predicting continuous numerical values.

$$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

### Classification: Cross-Entropy
Used for categorical tasks to measure the difference between two probability distributions.

> **Key Takeaway:** The loss function provides the numerical feedback the model needs to understand how 'wrong' its current predictions are.

<a name="part-8"></a>
## Part 8: Backpropagation

**Backpropagation** is the engine of neural network learning. It uses the **Chain Rule** from calculus to calculate how much each weight and bias contributed to the total error.

### The Learning Mechanism
After the loss is calculated, the signal is sent **backward** from the output layer to the input layer. The algorithm computes gradients, which indicate the direction and magnitude of change needed for each parameter.

> **Key Takeaway:** Backpropagation attributes the model's error to specific parameters, allowing the network to update itself systematically.

---
<br>

<a name="part-9"></a>
## Part 9: Gradient Descent

**Gradient Descent** is an optimization algorithm that adjusts the model's parameters to minimize the loss function.

### Optimization Logic
Think of the loss function as a valley. Gradient Descent helps the model 'walk downhill' toward the lowest point of error.

### The Learning Rate ($\alpha$)
The **Learning Rate** determines the size of the steps the model takes:
*   **Too high:** The model may overshoot the minimum and never converge.
*   **Too low:** The training will be extremely slow and might get stuck in a local minimum.

> **Key Takeaway:** Gradient Descent uses the gradients from backpropagation to update weights, effectively 'steering' the model toward high accuracy.

<a name="part-11"></a>
## Part 11: Building Your First Neural Network

Below is a practical implementation of a single-neuron model. We will train it to learn the linear relationship $y = 2x - 1$.

### Code Implementation
The following code demonstrates the modern Keras workflow:
1.  **Define Inputs:** Use the `Input` object to specify data shape.
2.  **Stack Layers:** Define the layers of the network.
3.  **Compile:** Choose an optimizer (like Stochastic Gradient Descent) and a loss function (MSE).
4.  **Fit:** Feed the data to the model to begin the learning process.

````python
import tensorflow as tf
import numpy as np
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense

# 1. Prepare data representing the formula: y = 2x - 1
xs = np.array([-1.0, 0.0, 1.0, 2.0, 3.0, 4.0], dtype=float)
ys = np.array([-3.0, -1.0, 1.0, 3.0, 5.0, 7.0], dtype=float)

# 2. Define the model architecture
# We use the explicit Input layer to define the shape (modern Keras standard)
model = Sequential([
    Input(shape=(1,)),       # Specifies that we receive 1 numerical input
    Dense(units=1)           # A single neuron to learn the linear relationship
])

# 3. Display the model summary to verify the architecture
model.summary()

# 4. Compile the model
# 'sgd' is Stochastic Gradient Descent; 'mean_squared_error' is our loss metric
model.compile(optimizer='sgd', loss='mean_squared_error')

# 5. Train the model (500 iterations over the dataset)
print("\nStarting training...")
model.fit(xs, ys, epochs=500, verbose=0)
print("Training complete.")

# 6. Make a prediction for a new value (x = 10)
prediction = model.predict(np.array([10.0]))
print(f"\nPrediction for input 10.0: {prediction[0][0]:.4f}")
print("Expected target value: 19.0")
````

## Part 13: Full ANN Project — Purchase Prediction

In this final section, we move from a single neuron to a multi-layered **Artificial Neural Network**. We will build a model that predicts whether a customer will purchase a product based on their **Age** and **Income**.

### Project Workflow
1.  **Data Creation:** Defining a small dataset using `pandas`.
2.  **Feature Engineering:** Separating independent variables ($X$) from the target ($y$).
3.  **Architecture Design:** Implementing a deep network with multiple hidden layers.
4.  **Training:** Optimizing the model using `Binary Crossentropy`.

````python
import pandas as pd
import tensorflow as tf
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense

# --- Step 1: Data Preparation ---
data = {
    "Age": [22, 25, 30, 35, 40, 28, 32, 45],
    "Income": [25000, 30000, 40000, 50000, 70000, 35000, 45000, 80000],
    "Purchased": [0, 0, 0, 1, 1, 0, 1, 1]
}

df = pd.DataFrame(data)

# --- Step 2: Feature Selection ---
X = df[["Age", "Income"]]
y = df["Purchased"]

# --- Step 3: Model Architecture ---
model = Sequential([
    Input(shape=(2,)),                 # Receive two inputs (Age and Income)
    Dense(16, activation="relu"),     # First hidden layer
    Dense(8, activation="relu"),      # Second hidden layer
    Dense(1, activation="sigmoid")    # Output layer for binary classification
])

# --- Step 4: Compilation ---
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# --- Step 5: Model Summary ---
model.summary()

# --- Step 6: Model Training ---
print("\nStarting the training process...")
model.fit(X, y, epochs=10, verbose=1)
print("\nTraining complete.")



# --- Step 7: Prediction on New Data ---
# Let's predict for a person: Age 33, Income 48,000
new_customer = np.array([[33.0, 48000.0]])

# The output will be a probability between 0 and 1 (due to sigmoid)
prob = model.predict(new_customer)

print(f"\nNew Prediction for Age 33, Income $48k: {prob[0][0]:.4f}")

if prob[0][0] > 0.5:
    print("Result: Likely to Purchase")
else:
    print("Result: Unlikely to Purchase")
````

> **Key Takeaway:** Real-world projects involve stacking multiple hidden layers with ReLU activations and using a Sigmoid output for binary classification to capture complex patterns between features like Age and Income.

Here's a brief overview of the layers in our `Sequential` model:

```python
model = Sequential([
    Input(shape=(2,)),                 # Two inputs: Age and Income

2.  **`Dense(16, activation="relu")`**: A fully connected hidden layer with 16 neurons. `ReLU` (Rectified Linear Unit) is used to introduce non-linearity, outputting the input if positive, else zero.

3.  **`Dense(8, activation="relu")`**: Another fully connected hidden layer, this time with 8 neurons, also using `ReLU` activation.

4.  **`Dense(1, activation="sigmoid")`**: The output layer, with a single neuron. `Sigmoid` activation is used to produce a probability between 0 and 1, suitable for binary classification (e.g., likelihood of purchase).
